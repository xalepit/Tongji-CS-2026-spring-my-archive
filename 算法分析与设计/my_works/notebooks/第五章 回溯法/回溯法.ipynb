# 基本定义
许多问题需要**找出它的所有解（解集）**，或者找出**满足某个约束条件的解**；

需要一步一步搜索：
```plain text
第1步选什么？
第2步选什么？
第3步选什么？
……
```
每一步有很多选择：
```plain text
做一个选择
↓
继续递归往后试（若成功，记录结果，返回）
↓
撤销选择
```
**深度优先**的**搜索**方法：这就是回溯法

# 适用场景
```plain text
“所有方案”
“所有组合”
“所有排列”
“棋盘排列”
“是否存在一种解/方案”
“路径搜索”
“满足某个约束条件”
```

# 相关定义
## 结点
- 扩展结点：当前正在生成子节点/搜索的结点
- 活结点：自身已被生成，但子节点/子树还在扩展的结点
- 死结点：所有子节点已被生成和扩展
## 解空间
- 解向量：回溯法问题的解可以表示成一个n元组$$(x_1,x_2,...,x_n)$$，每一步做的选择就是$$x_i$$ **（每一步都要枚举当前所有的可能选择）**
- 解空间：所有满足显式约束条件的解向量构成解空间
- 子集树：解空间是n个元素构成的集合的满足某个性质的某些子集
- 排列数：解空间是n个元素的满足某个性质的某些排列
- 回溯树：**每条边代表一种选择**
## 约束/剪枝
- 约束函数：要求解可行，用于剪不合法的解
- 限界函数：要求解有可能最优，比当前最优更优的，用于剪不可能更优的


# 通用模板：
递归回溯
```python
def backTrack(当前状态)
    if 到达临界/目标状态：
        记录答案
        return
    
    for each 当前的选择：
        if legal(当前选择):
            做出选择
            backTrack(新状态)
            撤销选择
```
