## 1.问题定义
- 有 n 个集装箱，第 i 个集装箱重量为 w[i]，有两艘轮船，载重量分别为 c1 和 c2。要求判断是否能把所有集装箱装上两艘船；如果能，找出一种装载方案。

- 思路：先尽可能把第一艘船装满，剩余集装箱再装第二艘船。因此问题转化为：从所有集装箱中选一个子集，使其总重量不超过 c1，且尽可能接近 c1。这是一个特殊的 0-1 背包问题

---

## 2.解法
- 解向量为：
$$
x = (x_1, x_2, ..., x_n)
$$
其中：
$$
x_i = 1 表示第 i 个集装箱装入第一艘船
x_i = 0 表示第 i 个集装箱不装入第一艘船
$$
-解空间：子集树
-每个集装箱都有两种选择：装/不装。所以搜索树是二叉树
搜索时维护变量：
```plain text
cw：当前第一艘船已装重量
bestw：当前找到的最优装载重量
r：剩余还没考虑的集装箱总重量
```
剪枝:
```plain text
约束函数：cw + x <= c1
限界函数: cw + r > bestw 
```
“当前重量 + 剩余所有重量”仍然可能超过当前最优解，就继续搜索；否则剪枝。

---

伪代码：
```python
Backtrack(i):
    如果 i > n：
        如果当前载重量 cw 大于 bestw：
            更新 bestw 和最优方案
        返回

    从剩余重量 r 中减去 w[i]

    尝试左子树：装入第 i 个集装箱
        如果 cw + w[i] <= c1：
            将 x[i] 置为 1
            cw 增加 w[i]
            递归搜索 Backtrack(i + 1)
            撤销选择，恢复 cw

    尝试右子树：不装入第 i 个集装箱
        如果 cw + r > bestw：
            将 x[i] 置为 0
            递归搜索 Backtrack(i + 1)
        否则剪枝（继续）

    恢复剩余重量 r （r, cw, bestw是函数外维护的变量）
```
