### 1.问题定义

在 n × n 的棋盘上放置 n 个皇后，使得任意两个皇后都不能互相攻击。也就是说，任意两个皇后不能位于：
```plain text
同一行
同一列
同一条斜线
```
由于每一行必须且只能放一个皇后，可以用 n 元组表示一个解。解向量为：
$$
x = (x_1, x_2, ..., x_n)
$$
其中：
$$
x_i 表示第 i 行皇后所在的列
$$
约束条件为：
$$
x_i != x_j \\
|i - j| != |x_i - x_j|
$$
也就是不能同列，不能同斜线。

### 2.回溯法解法
每一层放置一行的皇后。
第 i 层表示：决定第 i 行皇后放在哪一列
每次尝试一个列号 j，检查：
-该列是否已有皇后
-两条斜线上是否已有皇后

如果都没有冲突，则放置皇后，继续搜索下一行。

约束函数：当前皇后不能与之前已放置的皇后处于同一列或同一斜线。 

### 3.伪代码
```python
def backTrack(i):
    如果i > n:  
        记录一个合法放置方案
        返回
    
    for j in range(n): #枚举这一行所有可能的列
        if 与之前的冲突：
            continue #剪枝

        在该列放上皇后
        backTrack(i + 1)
        撤销皇后放置

```
